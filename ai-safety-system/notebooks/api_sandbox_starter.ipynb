Notebook provides examples for interacting with both the scenario APIs and the monitor endpoints.

Default base URL assumes the sandbox was started with:

```bash
docker run -d --rm -p 8030:8000 -p 8501:8501 -p 8502:8502 -p 8503:8503 --name comp40771-api-sandbox pedrombmachado/comp40771-api-sandbox:latest
```


````python
import json
import time
from pprint import pprint

import requests

BASE_URL = "http://localhost:8030"
````

````python
def api_call(method, endpoint, *, params=None, json_body=None, timeout=30):
    url = f"{BASE_URL}{endpoint}"
    response = requests.request(method, url, params=params, json=json_body, timeout=timeout)
    try:
        payload = response.json()
    except Exception:
        payload = response.text
    print(f"{method} {url} -> {response.status_code}")
    return response.status_code, payload


def show(method, endpoint, **kwargs):
    status, payload = api_call(method, endpoint, **kwargs)
    pprint(payload)
    return payload
````

## Polling monitor data while an API is running

````python
def poll_monitors(iterations=5, delay=2):
    rows = []
    for i in range(iterations):
        _, network = api_call("GET", "/monitor/network")
        _, cpu = api_call("GET", "/monitor/cpu")
        _, threads = api_call("GET", "/monitor/threads")
        _, storage = api_call("GET", "/monitor/storage")
        rows.append({
            "step": i + 1,
            "active_ports": network["summary"]["active_ports"],
            "total_packets_seen": network["summary"]["total_packets_seen"],
            "cpu_percent": cpu["cpu"]["system_cpu_percent"],
            "telemetry_threads": threads["threads"]["telemetry_worker_threads"],
            "file_storage_bytes": storage["storage"]["total_file_storage_bytes"],
        })
        time.sleep(delay)
    return rows

poll_monitors(iterations=3, delay=2)
````

````python
# Get the classification result for the most recent API run
show("GET", "/results/last")
````

````python
# Get the full history of API run results
show("GET", "/results")
````

## Helper functions for retrieving labels

````python
def get_last_result():
    return show("GET", "/results/last")


def get_all_results():
    return show("GET", "/results")


def get_latest_label():
    payload = get_last_result()
    return {
        "run_id": payload["run_id"],
        "api_id": payload["api_id"],
        "classification": payload["classification"],
        "potentially_harmful": payload["potentially_harmful"],
    }

get_latest_label()
````

## Example loop collecting labels from multiple API runs

````python
show("POST", "/apis/stop")
show("POST", "/telemetry/stop")
show("POST", "/telemetry/start", json_body={})

collected = []
for api_id in ["api01", "api02", "api03"]:
    show("POST", "/apis/stop")
    started = show("POST", f"/apis/{api_id}/start")
    result = show("GET", "/results/last")
    collected.append({
        "api_id": api_id,
        "run_id": started["run_id"],
        "classification": result["classification"],
        "potentially_harmful": result["potentially_harmful"],
    })

collected
````

## Example format for RL or downstream model training

````python
def collect_training_records(api_ids):
    records = []
    show("POST", "/apis/stop")
    show("POST", "/telemetry/stop")
    show("POST", "/telemetry/start", json_body={})
    for api_id in api_ids:
        show("POST", "/apis/stop")
        started = show("POST", f"/apis/{api_id}/start")
        label = show("GET", "/results/last")
        records.append({
            "api_id": api_id,
            "run_id": started["run_id"],
            "label": label["classification"],
            "potentially_harmful": label["potentially_harmful"],
        })
    return records

training_records = collect_training_records(["api04", "api05", "api06"])
training_records
````
